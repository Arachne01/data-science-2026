```python
import pandas as pd
import numpy as np
from scipy.stats.mstats import winsorize
```

```python
df = pd.read_csv("https://drive.google.com/uc?export=download&id=1LfQWProB0VjWN5q8bKuRIgn-stULfIRo")
```

```python
print(df.info)
```

```python
print(df.describe)
```

```python
print(df.isnull().sum())
```

```python
df.drop_duplicates(inplace=True)
print('Setelah hapus duplikat:', df.shape)
```

```python
df['kota'] = df['kota'].str.strip().str.title()
df['kondisi'] = df['kondisi'].str.strip().str.lower()
```

str.strip() menghilangkan spasi diawal maupun akhir dari inputan kolom kota, str.title() membuat format setiap inputan di kolom dimulai dengan huruf kapital dan diikuti oleh huruf kecil, str.lower() mengubah isi kolom menjadi semuanya huruf kecil

```python
df['luas_m2'] = df['luas_m2'].fillna(df['luas_m2'].median())
df['harga_juta'] = df['harga_juta'].fillna(df['harga_juta'].median())
df['kamar'] = df['kamar'].fillna(df['kamar'].mode()[0])
```

fillna(df[''].median()) berfungsi mengisi kekosongan dengan nilai median
fillna(df[''].mode()[0])berfungsi mengisi kekosongan dengan nilai modus

```python
for col in ['harga_juta', 'luas_m2', 'tahun_bangun']:
  Q1, Q3 = df[col].quantile([0.25, 0.75])
  IQR = Q3 - Q1
  df[col]= df[col].clip(Q1-1.5*IQR, Q3+1.5*IQR)
```

```python
assert df.isnull().sum().sum() == 0, 'Masih ada missing!'
assert df.duplicated().sum() == 0, 'Masih ada duplikat!'
print('Shape akhir : ', df.shape)
df.to_csv('housing_clean.csv', index=False)
print('Dataset bersih tersimpan!')
```
