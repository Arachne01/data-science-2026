```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
print(df.shape, df.columns.tolist())
```

Pertanyaan 1, terdapat 12 kolom dan juga 891 bars

```python
print(df.isnull().sum())
```

Pertanyaan 2, terdapat 177 values age, 687 values Cabin, dan 2 values Embarked yang masih kosong

```python
wk1 = df[(df['Sex']=="female") & (df['Pclass']==1)]

print(df['Survived'].mean(), wk1['Age'].mean())

```

Pertanyaan 3, ada 38,38% penumpang yang selamat
pertanyaan 4, Rata-rata usia penumpang wanita kelas 1 yang selamat adalah 34

```python
hasil = df.groupby('Pclass').agg(
    total=('PassengerId','count'),
    selamat=('Survived','mean'),).round(3)

print(hasil)
```

Pertanyaan 5, kelas 1 memiliki tingkat keselamatan yang paling tinggi
