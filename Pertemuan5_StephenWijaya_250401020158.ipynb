```python
import pandas as pd, numpy as np
import seaborn as sns
import matplotlib.pyplot as plt, matplotlib.gridspec as gridspec

df = sns.load_dataset('penguins')
print(df.shape, df.dtypes)
print(df.describe().round(2))

fig = plt.figure(figsize=(16, 12))
fig.suptitle('Dashboard Analisis Dataset Penguin',
             fontsize=18, fontweight='bold', y=0.98)

gs = gridspec.GridSpec(2, 2, figure=fig, hspace=0.4, wspace=0.35)

ax1=fig.add_subplot(gs[0,0])
ax2=fig.add_subplot(gs[0,1])
ax3=fig.add_subplot(gs[1,0])
ax4=fig.add_subplot(gs[1,1])

#ax1
data_species = df['species'].value_counts()
ax1.bar(data_species.index, data_species.values, color='teal', edgecolor='black')
ax1.set_title("Jumlah Penguin per spesies")

#ax2
df_sorted = df.dropna().sort_values(by='flipper_length_mm')
ax2.plot(df_sorted['flipper_length_mm'], df_sorted['body_mass_g'], color='coral', linewidth=2)
ax2.set_title("Hubungan Panjang Sirip dan Berat Badan")
ax2.set_xlabel("Panjang Sirip (mm)")
ax2.set_ylabel("Berat Badan (g)")
sns.lineplot(data=df_sorted, x='flipper_length_mm', y='body_mass_g', color='coral', ax=ax2, linewidth=2)

#ax3
sns.boxplot(data=df, x='species', y='body_mass_g',hue='species', ax=ax3, palette='Set1')
ax3.set_title("Distribusi Berat Badan per Spesies")
ax3.set_xlabel("Spesies Penguin")
ax3.set_ylabel("Berat Badan (g)")

#ax4
sns.scatterplot(data=df, x='flipper_length_mm', y='body_mass_g', hue='species', ax=ax4, palette='Set2', s=80)
ax4.set_title("Sebaran Panjang Sirip vs Berat Badan per Spesies")
ax4.set_xlabel("Panjang Sirip (mm)")
ax4.set_ylabel("Berat Badan (g)")


plt.savefig('dashboard_penguin.png', dpi=150, bbox_inches='tight')
plt.show()
print('Dashboard tersimpan sebagai dashboard_penguin.png')

```
